```python
df['Item']=df['Item'].fillna('Unknown')
```

```python
initial_row_count=len(df)
```

```python
df=df.drop_duplicates()
```

```python
final_row_count = len(df)
removed_duplicates_count = initial_row_count - final_row_count
```

```python
print(f"İlkin ümumi sətir sayı: {initial_row_count}")
print(f"Silindi təkrar sətir sayı: {removed_duplicates_count}")
print(f"Yekun sətir sayı: {final_row_count}")
```

```python
text_columns=['Item','Payment Method','Location']
```

```python
for col in text_columns:
    if col in df.columns:
      df[col] = df[col].astype(str).str.strip().str.title()
      df[col] = df[col].replace({'Nan': 'Unknown', 'None': 'Unknown'})
```

```python
if 'Gender' in df.columns:
    df['Gender'] = df['Gender'].str.strip().str.capitalize()
    df['Gender'] = df['Gender'].replace({'M': 'Male', 'F': 'Female'})
```

```python
if 'Transaction Date' in df.columns:
  df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], errors='coerce')
```

```python
print("Məlumatlar uğurla standartlaşdırıldı və datetime formatına keçirildi!")
```

```python
for col in ['Quantity', 'Price Per Unit', 'Total Spent']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
for col in ['Quantity', 'Price Per Unit', 'Total Spent']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col} sütununda outlier sayı: {len(outliers)}")
```

```python
for col in ['Quantity', 'Price Per Unit', 'Total Spent']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col} sütununda outlier sayı: {len(outliers)}")
```

```python
df['Transaction ID'] = df['Transaction ID'].astype(str)
df['Item'] = df['Item'].astype(str)
df['Quantity'] = df['Quantity'].astype(float)
df['Price Per Unit'] = df['Price Per Unit'].astype(float)
df['Total Spent'] = df['Total Spent'].astype(float)
df['Payment Method'] = df['Payment Method'].astype(str)
df['Location'] = df['Location'].astype(str)
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], errors='coerce')
```

```python
df.to_csv('cleaned_cafe_sales.csv', index=False)

print("Fayl uğurla yadda saxlanıldı!")
```
